```python
import pandas as pd

# Carregar um CSV pra ver a estrutura
df = pd.read_csv('../data/raw/reclamacoes_2024_T1.csv', sep=';', encoding='latin-1')
df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
bancos = [
    'Nubank', 'C6', 'Inter', 'Itaú', 'Bradesco', 
    'Santander', 'Banco do Brasil', 'PagBank', 'Mercado Pago', 'PicPay'
]

instituicoes = df['Instituição financeira'].unique()
for banco in bancos:
    match = [i for i in instituicoes if banco.lower() in i.lower()]
    print(f"{banco}: {match}")
```

```python
for termo in ['itau', 'brasil', 'picpay']:
    match = [i for i in instituicoes if termo.lower() in i.lower()]
    print(f"{termo}: {match}")
```

```python
for termo in ['bb ', 'banco do brasil', 'bco brasil', 'picpay', 'pic pay']:
    match = [i for i in instituicoes if termo.lower() in i.lower()]
    print(f"{termo}: {match}")
```

```python
# Verifica em qual trimestre o PicPay aparece pela primeira vez
import os

arquivos = sorted(os.listdir('../data/raw/'))
# Para cada arquivo, carrega e busca PicPay
for arquivo in arquivos:
    df_temp = pd.read_csv(f'../data/raw/{arquivo}', sep=';', encoding='latin-1')
    instituicoes_temp = df_temp['Instituição financeira'].unique()
    match = [i for i in instituicoes_temp if 'picpay' in i.lower()]
    print(f"{arquivo}: {match}")
```

```python
# Define o mapeamento dos bancos
mapeamento = {
    'Nubank': 'NUBANK (conglomerado)',
    'C6': 'C6 BANK (conglomerado)',
    'Inter': 'INTER (conglomerado)',
    'Itaú': 'ITAU (conglomerado)',
    'Bradesco': 'BRADESCO (conglomerado)',
    'Santander': 'SANTANDER (conglomerado)',
    'Banco do Brasil': 'BB (conglomerado)',
    'PagBank': 'PAGBANK-PAGSEGURO (conglomerado)',
    'Mercado Pago': 'MERCADO PAGO (conglomerado)',
    'PicPay': 'PICPAY (conglomerado)'
}

# Filtra só os 10 bancos e verifica nulos no Índice
nomes_bacen = list(mapeamento.values())
df_bancos = df[df['Instituição financeira'].isin(nomes_bacen)]
print(df_bancos[['Instituição financeira', 'Índice']].to_string())
```

```python
# Verifica se todos os arquivos têm as mesmas colunas
colunas_por_arquivo = {}

for arquivo in arquivos:
    df_temp = pd.read_csv(f'../data/raw/{arquivo}', sep=';', encoding='latin-1')
    colunas_por_arquivo[arquivo] = df_temp.columns.tolist()

# Compara tudo com o primeiro arquivo
colunas_base = colunas_por_arquivo[arquivos[0]]
for arquivo, colunas in colunas_por_arquivo.items():
    if colunas != colunas_base:
        print(f"DIFERENTE: {arquivo}")
        print(colunas)
    else:
        print(f"OK: {arquivo}")
```
